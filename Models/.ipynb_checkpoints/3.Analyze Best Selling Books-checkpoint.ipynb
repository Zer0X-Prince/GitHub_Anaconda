```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\frien\Downloads\bestsellers with categories.csv")
df.head()
```

```python
print(df.columns) # list of columns 
print(df.shape) # (rows, columnns)
```

```python
df.describe()
```

```python
df.info()
```

```python
# checking null values
df.isnull().sum()
```

```python
# removing duplicates rows
df.drop_duplicates(inplace=True) # by using inplace=true changes are made directly to the original DataFrame
```

```python
df.head()
```

```python
df.columns
```

```python
# renaming columns 
df.rename(columns = {'Name': 'Title', 'Year': 'Publication Year', 'User Rating': 'Rating'}, inplace = True)
df.columns
```

```python
df.head()
```

```python
# coverting data types 
df['Price'] = df['Price'].astype(float)
```

```python
df.head()
```

```python
# analyzing author popularity
author_counts = df['Author'].value_counts()
author_counts.head(10)
```

```python
df['Genre'].value_counts()
```

```python
ave_rating_by_genre = df.groupby('Genre')['Rating'].mean()
ave_rating_by_genre
```

```python
df.groupby('Genre')['Rating'].sum()
```

```python
df[df['Rating']>=4].groupby('Genre')['Rating'].value_counts().groupby('Genre').sum()
```

```python
df.groupby('Genre')['Rating'].value_counts()
```

```python
# exporting to csv file avg raiting and top author 
ave_rating_by_genre.to_csv('ave_rating_by_genre.csv')
author_counts.head(10).to_csv('top_author_counts.csv')
```
