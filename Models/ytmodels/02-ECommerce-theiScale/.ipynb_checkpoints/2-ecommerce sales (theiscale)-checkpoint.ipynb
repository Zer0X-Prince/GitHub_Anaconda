````python
# converting data type
data['Order Date'] = pd.to_datetime(data['Order Date'])
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
````

````python
data.info()
````

````python
data.head()
````

````python
# creating new columns for month, year, day of week
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
````

### Monthly Sales Analysis
Q1. Calculate monthly sales and identify which month had highest and lowest sales

<details><summary>Click here for the solution</summary>
highest sales = nov (11)<br>
lowest sales = jan (1)

</details>

````python
# monthly sales analysis
sales_by_month = data.groupby('Order Month')['Sales'].sum().reset_index() # reset_index() add new indexing
sales_by_month
````

````python
# ploting using plotly
flg = px.line(sales_by_month,
              x = 'Order Month',  # column name
              y = 'Sales', # column name
              title = 'Monthly Sales Analysis')
flg.show()
````

### Sales By Categories
Q2. Analyze sales based on product categories and detemine which category has lowest and highest sales

<details><summary>Click here for the solution</summary>
highest sales = technology <br>
lowest sales = office suplies

</details>

````python
data.head()
````

````python
data.columns
````

````python
sales_category = data.groupby('Category')['Sales'].sum().reset_index()
sales_category
````

````python
# line graph
flg2 = px.line(sales_category,
    x = 'Category',
    y = 'Sales',
    title = 'Sales by Category')
flg2.show()
````

````python
# pie chart
flg2 = px.pie(
    sales_category,
    names='Category',   # replaces x
    values='Sales',     # replaces y
    title='Sales by Category',
    hole = 0.3 # for hole in middle
    # color_discrete_sequence = px.colors.qualitatve.Pestel
)
flg2.update_traces(textposition= 'inside', textinfo= 'percent + label')
flg2.update_layout(title_text= 'Sales Analysis by Category')

flg2.show()
````

### Sales Analysis By Sub-Categories
Q3. Sales analysis based on sub-categories

````python
data.columns
````

````python
sales_subcat = data.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_subcat
````

````python
# pie chart
flg3 = px.pie(sales_subcat,
              names = 'Sub-Category',
              values = 'Sales',
              title = 'Sales Analysis By Sub-Categories',
              hole = 0.2)
flg3.update_traces(textposition = 'inside',textinfo = 'percent + label')
flg3.update_layout(width = 900, height = 700)
flg3.show()
````

````python
# bar graph
flg3 = px.bar(sales_subcat,
              x = 'Sub-Category',
              y = 'Sales',
              title = 'Sales Analysis By Sub-Categories')
flg3.show()
````

### Sales And Profit By Customer Segment
Q6. Analyze sales and profit by customer segment

````python
data.columns
````

````python
sales_customer_segment = data.groupby('Segment')['Sales'].sum().reset_index()
profit_customer_segment = data.groupby('Segment')['Profit'].sum().reset_index()
profit_customer_segment
````

````python
customer_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
customer_segment
````

````python
flg6 = px.bar(
    customer_segment,
    x='Segment',
    y=['Sales', 'Profit'],   # multiple columns
    barmode='group',         # side-by-side bars
    title='Sales and Profit by Customer Segment'
)

flg6.update_layout(width=900, height=600)

flg6.show()
````

### Sales To Profit Ratio
Q7. Sales to profit ratio

````python
customer_segment = data.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
customer_segment['sale_to_profit_ratio'] = customer_segment['Sales'] / customer_segment['Profit']
customer_segment[['Segment','sale_to_profit_ratio']]
````
