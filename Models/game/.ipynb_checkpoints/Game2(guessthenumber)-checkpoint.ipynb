```python
import random
n = random.randint(0,100)
guesses = 1
a = -1
while (a != n):
    a = int(input('Guess the number : ' ))
    if a < n :
        print('higher number please ')
        guesses += 1
    elif a > n:
        print('lower number please')
        guesses += 1
print(f'You have guessed the number {n} correctly in {guesses} attempts')
```
