```python
import string
from collections import Counter

class TextAnalyzer:
    
    def __init__(self, text):
        self.text = text.lower()
        
        # Remove punctuation
        for p in string.punctuation:
            self.text = self.text.replace(p, "")
        
        self.words = self.text.split()
    
    
    def word_count(self):
        return len(self.words)
    
    
    def sentence_count(self):
        return self.text.count('.') + self.text.count('!') + self.text.count('?')
    
    
    def unique_words(self):
        return len(set(self.words))
    
    
    def most_common_words(self, n=5):
        return Counter(self.words).most_common(n)


# ---- Using the Class ----
text = """Data science is amazing. Data science uses Python.
Python makes data analysis easy and powerful."""

analyzer = TextAnalyzer(text)

print("Total Words:", analyzer.word_count())
print("Unique Words:", analyzer.unique_words())
print("Most Common Words:", analyzer.most_common_words())
```
