# String

````python
name = "Education"
````

````python
# String Slicing = str[start: End: Stepsize]
name[0:4:2]
````

````python
# String Functions
print(len(name))
print(name.upper())
print(name.lower())
print(f'my name is python and {name}'.title())
print(F' length of {name} =', len(name))
print(f' replacing t of {name} as X =', name.replace("t", 'X'))
print(name.find('i'))
print(name.find('l')) # -1 if not present
print(name.endswith("ion"))
print(name.count("u"))
print(name.capitalize())
````

````python
# escape sequence \n(next line), \t(tab space), \\(slash), \'(comma)
print(f'my \' name \n is \t python \\ and {name}'.capitalize())
````

# List

````python
l1 = [1,2,3,'a','b']
l2 = [1,2,3]
l3 = [4,5,6]
l4 = [45,54,34,00,33,1]
````

````python
print(l1[2])
print(l1[3])
print(l1[1:6]) #list slicing same as string slicing l1[start: end: stepsize]
````

````python
l4.sort() # only for numeric values
print(l4)
l4.sort(reverse = True) # desending 
print(l4)
````

````python
l1.reverse() # reverse values
print(l1)
````

````python
# list methods
l2.append('x') # add at last index
l2.insert(1, 'y') # add at position metioned
l5 = l2 + l3 # concatenation
print(l2)
print(l5)
````

````python
# adding multiple values
l7 = [1,2,3]
l7.extend([22,33,44,"a", 'b'])
l7
````

````python
nested_list = [l1, l2, l3]
print(nested_list)
````

````python
# try and except
a = 1
try:
    b = int(input("Please enter a number to divide a"))
    a = a/b
    print("Success a=",a)
except:
    print("There was an error")
````

<details>
    <summary> SYNTAX of try--except--except--except--else</summary>
    
```python
# potential code before try catch

try:
    # code to try to execute
except ZeroDivisionError:
    # code to execute if there is a ZeroDivisionError
except NameError:
    # code to execute if there is a NameError
except:
    # code to execute if ther is any exception
    
# code that will execute if there is no exception or a one that we are handling
```
</details>

````python
# Example
a = 1
try:
    b = int(input("Please enter a number to divide a"))
    a = a/b
    print("Success a=",a)
except ZeroDivisionError:
    print("The number you provided cant divide 1 because it is 0")
except ValueError:
    print("You did not provide a number")
except:
    print("Something went wrong")
````

<details>
    <summary>SYNTAX of try--except--except--except--else--finally</summary>

```python
# potential code before try catch

try:
    # code to try to execute
except ZeroDivisionError:
    # code to execute if there is a ZeroDivisionError
except NameError:
    # code to execute if there is a NameError
except:
    # code to execute if ther is any exception
else:
    # code to execute if there is no exception
finally:
    # code to execute at the end of the try except no matter what
    
# code that will execute if there is no exception or a one that we are handling
```
</details>

````python
# Example
a = 1

try:
    b = int(input("Please enter a number to divide a"))
    a = a/b
except ZeroDivisionError:
    print("The number you provided cant divide 1 because it is 0")
except ValueError:
    print("You did not provide a number")
except:
    print("Something went wrong")
else:
    print("success a =",a)
finally:
    print("Processing Complete")
````

````python
# question 1
def devide(numer, denomer):
    try:
        result = numer/denomer
        return result
    except ZeroDivisionError:
        print('Cannot devide by 0')

a = int(input('enter numerater: '))
b = int(input('enter denomintor: '))
devide(a, b)
````

````python
# qeustion 2
import math
def sqrt(num):
    try :
        num1 = math.sqrt(num)
        return num1
    except : 
        print('Invalid No. please enter positive no.')
a = int(input('enter no. '))
sqrt(a)
````

````python
# qeustion 3
def complex_calculation(num):
    try:
        result = num / (num - 5)
        print (f"Result: {result}")
    except Exception as e:
        print("An error occurred during calculation.")
# Test case
user_input = float(input("Enter a number: "))
complex_calculation(user_input)
````
