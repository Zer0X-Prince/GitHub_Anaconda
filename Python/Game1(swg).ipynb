```python
import random
            
computer = random.choice([1, 0, -1])

user = input('Enter your choice from (s = snake | w = water | g = gun)').lower()

dict = { 1 : 'snake', 0 : 'water', -1 : 'gun' }
reversedict = {'s' : 1, 'w' : 0 , 'g' : -1}
you = reversedict[user]

print(f'Computer choose {dict[computer]} and you choose {dict[you]}')
             
if computer == you : 
    print('It\'s Draw!')
elif computer - you == -1 or computer - you == 2 : 
    print('You Win!')
else : 
    print('You Loose!')
```
