#### Array Operations

```python
# 1D array
a = np.array([1,3,5])
b = np.array([2,4,6])
#2D array
a2 = np.array([[1,3,5],[2,4,6]])
b2 = np.array([['a','b','c'],[1,2,3]])
```

```python
# elements wise operations
print('addtion = ', a + b)
print('multiplication = ', a * b)
print('squared = ', a **2)
print('square root = ', np.sqrt(a + b))
print('scaler multiplication = ', a * 5)
```

```python
a2
```

```python
# broadcasting
print('scaler multi = ', a2* 5)
print('scaler add = ', a2 + 5)
```

#### Case Study - insurance claims record

```python
claims = np.array([[12000,15000,18000],
                  [10000,13000,17000],
                  [9000,11000,16000]])
claims
```

```python
print(claims[0,2])
print(claims[2,0])
```

```python
print(np.sum(claims, axis = 1)) # row wise
print(np.sum(claims, axis = 0)) # columns wise
```

```python
print(np.mean(claims, axis = 1)) # row wise
print(np.mean(claims, axis = 0)) # columns wise
```

```python
print(np.sum(claims)) # total sum
```

```python
print(claims[claims>1300])
```

#### Case Study 2

```python
import numpy as np 
```

```python
data = np.array([
    ['C001', 5, 'Excellent'],
    ['C002', 4, 'Good'],
    ['C003', 3, 'Average'],
    ['C004', 5, 'Loved'],
    ['C005', 2, 'Poor'],
    ['C006', 4, 'Good'],
    ['C007', 1, 'Terrible'],
    ['C008', 5, 'Amazing'],
])
data
```

```python
# extracting only rating as int
ratings = data[:, 1].astype(int) # convert it into int values
len(ratings)
```

```python
# avg ratings
print(ratings.mean())
print(ratings.max())
print(ratings.min())
```

```python
feedbacks = data[:, 2].astype(str)
feedbacks
```

```python
unique_ratings, rating_counts = np.unique(ratings, return_counts=True)

print("Unique ratings:", unique_ratings)
print("Rating counts:", rating_counts)
```

```python
# total no of rating count per unique rating
for i,x in zip(unique_ratings, rating_counts):
    print(f' Customer ratings : {i} -> total rating count is {x}')
```

```python
# feedbacks
for c,r,comment in zip(data[:,0], ratings,data[:, 2]):
    print(f'Customer number {c} and feedback with ratings {r} is : {comment}')
```
