#### Handling missing values, aggregation, sorting

1 imputation -> fill missing values with mean, median, mode <br>
2 Deletion -> remove missing columns values

| Method | When to use |
|--------|-------------|
| Mean | numeric dt, normal distribution |
| Median | numeric dt, skewed distribution |
| Mode | catagorical dt ( eg : 'weather' in our dataset) |
| Delete | <5% missing dt and no meaningful pattern (MCAR - missing completely at random) |
