```python
import pandas as pd
```

```python
dict = {'city': ['Brooklyn', 'Seoul', 'Barcelona', 'Mexico City'],
        'country' : ['US', 'South Korea', 'Spain', 'Mexico'],
        'population' : [2646000, 9411000, 1636000, 9206944]}
```

```python
df = pd.DataFrame(dict, index = ['a', 'b', 'c', 'd'])
df
```

```python
# access columns
df[['city', 'population']]
```

```python
# accessesing elements with there index and colomns name using loc
df.loc['a','city']
```

```python
# accessesing elements with numeric postion using iloc
df.iloc[0 ,0]
```

```python
# df slicing
df.iloc[0:2 ,0:2]
```
